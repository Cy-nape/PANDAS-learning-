pyjanitor is a Python package providing convenient data cleaning methods extending pandas DataFrames with clean method-chaining syntax.

```python
from io import StringIO
import janitor
import numpy as np
import pandas as pd
```

### 1. Setting Header from a Data Row (row_to_names)
Real-world CSV or Excel files often contain messy leading rows, metadata, or notes before the actual table header appears.pd.read_csv(..., header=None) loads all rows as regular data records.   df.row_to_names(row_number=..., remove_row=True) elevates a specific row index to become the DataFrame column headers and optionally strips that row from the data.   

```python
# Raw simulated CSV text with extra rows before the actual headers
data = """shoe, 220, 100
shoe, 450, 40
item, retail_price, cost
shoe, 200, 38
bag, 305, 25"""

# Read into DataFrame without inferring headers
df_headers = pd.read_csv(StringIO(data), header=None)
df_headers
```

```python
# Promote index 2 ('item, retail_price, cost') to column names and remove it from rows
df_headers = df_headers.row_to_names(row_numbers=2, remove_row=True)
df_headers
```

## 2. Generating Combinations (janitor.expand_grid)
janitor.expand_grid() creates a Cartesian product of dictionary sequences (similar to R's expand.grid function). This is useful for grid searches, testing factor combinations, or scenario modeling.

```python
data = {"x": [1, 2, 3], "y": [1, 2]}

df_grid = janitor.expand_grid(others=data)
df_grid
```

## 3. Conditional / Range Joins (conditional_join)Standard 
pandas .merge() primarily supports exact equality (==). conditional_join allows non-equi joins using inequality conditions (such as >, >=, <, <=) across columns between two DataFrames.   
In this example:
    Match rows where df1['id'] == df2['id'].   
    Match rows where df1['value_1'] >= df2['value_2A'].   
    Match rows where df1['value_1'] <= df2['value_2B'] (i.e., value_1 falls between value_2A and value_2B). 

```python
df1 = pd.DataFrame(
    {
        "id": [1, 1, 1, 2, 2, 3],
        "value_1": [2, 5, 7, 1, 3, 4],
    }
)

df2 = pd.DataFrame(
    {
        "id": [1, 1, 1, 1, 2, 2, 2, 3],
        "value_2A": [0, 3, 7, 12, 0, 2, 3, 1],
        "value_2B": [1, 5, 9, 15, 1, 4, 6, 3],
    }
)

result = df1.conditional_join(
    df2,
    ("id", "id", "=="),
    ("value_1", "value_2A", ">="),
    ("value_1", "value_2B", "<="),
)
result
```

## 4. Transforming Columns in Method Chains (transform_column)
transform_column() applies a custom function or numpy operation directly to a target column. Because it returns the DataFrame directly, it simplifies multi-step pandas pipelines without requiring multiple reassignment lines.

```python
df_transform = pd.DataFrame({"values": [-1, -2, -3, -4, 5, 6]})
df_transform
```

```python
# Apply np.abs (absolute value) directly to the "values" column
df_transform = df_transform.transform_column("values", np.abs)
df_transform
```
