```python
import matplotlib.pyplot as plt
```

```python
x = [3,4,5,6,7,8]
y = [4,6,6,7,4,4]
plt.plot(x,y)
plt.show
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,y , color='red' , marker = 'o' , linestyle = '--', linewidth=2, markersize = 2)
plt.title("yehi h jo hai")
plt.xlabel("ye x hai ")
plt.ylabel("ye y hai ")
plt.show()
```

```python
x = [3,4,5,6,7,8]
y1 = [4,6,8,10,14,16]
y2 = [14,16,18,20,24,26]

plt.plot(x,y1, label = "y1 wala data")
plt.plot(x, y2, label = "y2 wala data")
plt.legend()
plt.show()
```
