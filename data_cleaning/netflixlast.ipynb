here we found that rating ki 3 rows me time hai like - 74 , 84 and 66 minutes ans duration me 3 values bhi missing hai - matlab ye valuese vahi ki h jo yaha kisi ne likh di hai 

```python
data["rating"].value_counts()
```

lekin hum ase direct thodi man saket h ki ye duration ki values hai or duration khali h in rows me, to hum ye specific rows ko dhekte hai 

```python
shifted_rows = data[data["rating"].isin(["74 min", "84 min" , "66 min"])]
```

here we can see clearly that duration khali hai or duration wali values rating me hai 

```python
shifted_rows
```

now we have to ship those value to the actual duration column 

```python
shiftmask = data["rating"].isin(["74 min", "84 min" , "66 min"])
```

```python
data.loc[shiftmask , "duration"] = data.loc[shiftmask , "rating"]
data.loc[shiftmask , "rating"] = None
```

```python
data.loc[shiftmask]
```

```python
data.groupby("type")["rating"].value_counts()
```

```python
data["rating"] = data["rating"].fillna(data["rating"].mode()[0])
```

```python
data.isna().sum()
```

```python
data["director"] = data["director"].apply(lambda x : "Unknown" if pd.isna(x) else x)
```

```python
data.isna().sum()
```

```python
for col in ["cast","country"]:
    data[col] = data[col].fillna("Unknown")
```

```python
data.isna().sum()
```

```python
data["date_added"] = data["date_added"].str.strip()
```

```python
data["date_added"] = pd.to_datetime(data["date_added"], format="%B %d, %Y")
```
